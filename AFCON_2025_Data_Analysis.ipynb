
# AFCON 2025 – Analyse des données  



Les bibliothèques principales utilisées sont :

pandas : manipulation et nettoyage des données

matplotlib / seaborn : visualisation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```


## Chargement des données
Les données sont stockées dans des fichiers CSV séparés (équipes , joueurs et matches).


```python

teams1 = pd.read_csv("data/Teams/afcon-teams-2025.csv")
teams2 = pd.read_csv("data/Teams/afcon-teams2-2025.csv")
players = pd.read_csv("data/Players/afcon-players-2025.csv")
matches = pd.read_csv("data/matches/afcon-matches-2025.csv")
players.head()
```


## Fusion des données des équipes
La fusion est réalisée à l’aide de la colonne `common_name`.


```python

teams_combined = pd.merge(
    teams1,
    teams2,
    on="common_name",
    suffixes=("", "_extra")
)

teams_combined.head()
```


## Nettoyage des données des joueurs
On conserve uniquement les joueurs ayant joué au moins une minute.


```python

players = players[players["minutes_played_overall"] > 0].copy()
players.shape
```


## Gestion des valeurs manquantes
Pour garantir la fiabilité de nos calculs, nous effectuons ici un remplissage des données manquantes. Nous remplaçons les valeurs vides par zéro pour les statistiques comme les buts ou les cartons, puis nous vérifions que notre tableau est parfaitement propre avant de lancer l'analyse.

```python

cols_to_fill = [
    "goals_overall",
    "assists_overall",
    "yellow_cards_overall",
    "red_cards_overall"
]

players[cols_to_fill] = players[cols_to_fill].fillna(0)
players[cols_to_fill].isna().sum()
```


## Indicateurs de performance


## Score d’impact
Un score est calculé à partir des buts et des passes décisives.


```python
# 1. Calcul du Score d'Impact
# Formule : Impact = (Buts * 0.6) + (Passes * 0.4)
players["impact_score"] = (
    players["goals_overall"] * 0.6 + 
    players["assists_overall"] * 0.4
)

# 2. Liste des colonnes "importantes" à afficher
important_features = [
    "full_name",  
    "goals_overall", 
    "assists_overall", 
    "impact_score"
]

# 3. Tri et affichage du Top 10
top_10_impact = players.sort_values("impact_score", ascending=False).head(10)

# On affiche uniquement le sous-ensemble de colonnes choisies
top_10_impact[important_features]
```

### Analyse des statistiques avancées (xG)

Cette partie du code vise à évaluer la dangerosité offensive des joueurs ainsi que leur capacité de finition, en comparant les buts réellement marqués aux buts attendus (Expected Goals – xG).

```python

xg_col = 'xg_total_overall'

players['clinical_finish'] = players['goals_overall'] - players[xg_col]

offensive_threat = players.nlargest(10, xg_col)[['full_name', 'goals_overall', xg_col, 'clinical_finish']]

print(" Analyse de la tendance offensive (xG) et de l’efficacité de finition :")
display(offensive_threat)
```

La variable xg_total_overall représente le nombre total de buts attendus pour chaque joueur.

Un indicateur appelé clinical_finish est calculé comme la différence entre :

les buts réellement marqués (goals_overall)

les buts attendus (xg_total_overall)


## Analyse de corrélation avec Scatter Plot (nuage de points) accompagné d'une ligne de régression.
Relation entre possession moyenne et nombre de victoires.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Calcul du coefficient de corrélation de Pearson (r)
# Ce chiffre indique la force du lien entre les deux variables
corr_value = teams_combined["average_possession"].corr(teams_combined["wins"])

# 2. Création du graphique
plt.figure(figsize=(12, 7))
sns.regplot(data=teams_combined, x="average_possession", y="wins", 
            scatter_kws={"color": "royalblue", "s": 100, "alpha": 0.7}, 
            line_kws={"color": "red", "label": f"Ligne de tendance (r={corr_value:.2f})"})

# 3. Ajout des noms des pays pour identifier les équipes
for i in range(teams_combined.shape[0]):
    plt.text(teams_combined.average_possession[i] + 0.3, 
             teams_combined.wins[i], 
             teams_combined.common_name[i], 
             fontsize=9)

# 4. Habillage du graphique
plt.title("Corrélation entre Possession et Victoires - CAN 2025", fontsize=15)
plt.xlabel("Possession Moyenne (%)", fontsize=12)
plt.ylabel("Nombre de Victoires", fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
corr_value
```

## Discipline et impact négatif sur la performance

### Objectif

Analyser le comportement disciplinaire des joueurs et évaluer son impact potentiel sur la performance collective de l’équipe.

### Indicateurs utilisés

Nombre de cartons jaunes

Nombre de cartons rouges

```python
# Indice disciplinaire pondéré
players["discipline_score"] = (
    players["yellow_cards_overall"] +
    3 * players["red_cards_overall"]
)

# Discipline par 90 minutes
players["discipline_per_90"] = players["cards_per_90_overall"]

# Top 10 des joueurs les plus sanctionnés
top_discipline = players.nlargest(
    10,
    "discipline_score"
)[[
    "full_name",
    "position",
    "yellow_cards_overall",
    "red_cards_overall",
    "discipline_score",
    "discipline_per_90"
]]

top_discipline
```

### le "Fair-Play" ou l'agressivité des équipes. Pour les équipes avec (Bar Chart)
 un carton rouge pèse bien plus lourd (3 points) qu'un carton jaune (1 point).

```python
import matplotlib.pyplot as plt

# 1. Calcul de l'indice disciplinaire pour les équipes
# Formule : Score = Jaunes + (3 * Rouges)
teams_combined["discipline_score"] = (
    teams_combined["cards_total"] + 
    2 * teams_combined["total_goal_count"] # Note: Ajustez selon vos colonnes de cartons rouges si disponibles
) 

# Si vous avez des colonnes spécifiques pour les rouges :
# teams_combined["discipline_score"] = teams_combined["yellow_cards_total"] + 3 * teams_combined["red_cards_total"]

# 2. Trier les données (les plus sanctionnés en haut)
df_discipline = teams_combined.sort_values(by="discipline_score", ascending=True)

# 3. Création du graphique
plt.figure(figsize=(12, 8))
colors = plt.cm.YlOrRd(df_discipline["discipline_score"] / df_discipline["discipline_score"].max())

plt.barh(df_discipline["common_name"], df_discipline["discipline_score"], color=colors)

# 4. Personnalisation
plt.title("Indice Disciplinaire par Équipe (Jaunes + 3*Rouges)", fontsize=15, fontweight='bold')
plt.xlabel("Score d'Indiscipline", fontsize=12)
plt.ylabel("Équipe", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Interprétation

Valeur élevée : joueur à risque
Un excès de sanctions peut perturber la stabilité tactique, provoquer des suspensions et affaiblir l’équipe.

## Analyse de la régularité des performances

### Objectif

Identifier les joueurs les plus constants tout au long du tournoi, indépendamment des performances ponctuelles.

### Indicateur de performance

average_rating_overall

goals_per_90_overall

assists_per_90_overall

minutes_played_overall


```python
# 1. Définir un seuil de minutes (ex: 270 minutes)
seuil_minutes = 270

# 2. Filtrer AVANT le calcul du score
regularity_ds = players[players["minutes_played_overall"] >= seuil_minutes][[
    "full_name",
    "position",
    "minutes_played_overall",
    "goals_per_90_overall",
    "assists_per_90_overall",
    "average_rating_overall"
]].dropna(subset=["average_rating_overall"])

# 3. Calcul du score de régularité
# La division par .max() sera maintenant basée sur les joueurs "valides"
regularity_ds["regularity_score"] = (
    regularity_ds["average_rating_overall"] * 0.5 +
    (regularity_ds["minutes_played_overall"] / regularity_ds["minutes_played_overall"].max()) * 0.3 +
    (regularity_ds["goals_per_90_overall"] + regularity_ds["assists_per_90_overall"]) * 0.2
)

# 4. Classement final
most_consistent = regularity_ds.sort_values(
    by="regularity_score", 
    ascending=False
).head(10)

most_consistent
```
